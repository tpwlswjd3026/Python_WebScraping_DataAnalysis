# 1. 뉴스 제목 추출하기

```python
#요청 파라미터를 설정하는 부분 sid는 섹션 id를 의미함
req_param = {
    'sid' : 105
}

#url을 만들고, sid 자리에 요청 파라미터를 넣어준다.
url = 'https://news.naver.com/section/{sid}'.format(**req_param) #딕셔너리를 풀어서 넣어준다

#요청 헤더를 설정하기 : 이런 브라우저에서 접속하는 거라고 알려주는 부분
req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}

# requests 의 get()함수를 호출하기
res = requests.get(url,headers=req_header) #url에 요청 헤더를 넣어서 요청을 보내고, 답은 res에 저장함
print(res.status_code, res.ok) #응답코드와 응답이 정상인지 확인함
 #200이면 성공, 404면 실패, 500이면 서버에러, 403이면 접근거부, 400이면 잘못된 요청

if res.ok: #요청이 성공했으면
    html = res.text #응답받은 html 문서를 문자열로 가져온다
    soup = BeautifulSoup(html, 'html.parser') #html 문서를 객체로 바꿔주고, 태그를 쉽게 찾을수 있다
    #이때 html.parser는 파이썬 내장 파서로 파싱에 사용한다.
    
    print(len(soup.select("div.sa.text a[href*='https://news.naver.com']"))) # 클래스가 sa.text인 div태그, 그 안에 <a>태그를 전부 찾음, 이때 len은 몇개가 검색되는지 개수 알려줌
    a_tags = soup.select("div.sa.text a[href*='https://news.naver.com']") # 위에서 찾은 태그를 a_tags라는 변수에 저장한다
    print(type(a_tags)) # a_tags의 자료형을 확인한다.

    #a_tags 리스트에 담긴 a_tag에서 꺼내면서 반복한다:태그 리스트 순회하는것
    for a_tag in a_tags:
        title = a_tag.get_strip() # <a>태그 안에 있는 텍스트 가져옴, strip은 공백 제거
        link = a_tag.get('href') #a_tag에서 기사 제목 부분을 가져옴
        print(title, link) #제목과 링크를 출력한다
    else:
        print(f'Error Code = {res.status_code}') #정상 응답이면 200, 아니면 에러코드줌)
```

## 1.1 뉴스 제목 추출하는 함수 선언하기

```python
import requests
from bs4 import BeautifulSoup #굳이 다시 안불러도 된다고 하는데 기억할겸 또 불러옴

#섹션의 키,값을 딕셔너리로 만들기
section_dict = {'정치':'100', '경제':'101', '사회':'102', '생활':'103', '문화':'103','생활/문화':'103', '세계':'104', 'IT':'105', '과학':'105','IT/과학':'105'}

def print_news(section_name):
    sid = section_dict.get(section_name,'102')
    url = f'https://news.naver.com/section/{sid}' #변수명 자리에 값이 들어가도록 f-string을 사용함
    print(f'{section_name} 뉴스 {url}') 

    #요청 헤더 설정하기
    req_header = {
            'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
        }

    # url로 req_header를 넣어서 요청을 보내고, 응답을 res에 저장함
    res = requests.get(url, headers=req_header)
    if res.ok:
        html = res.text #응답받은 html 문서를 문자열로 가져온다
        soup = BeautifulSoup(html, 'html.parser') #html 문서를 객체로 바꿔주고, 태그를 쉽게 찾을수 있다
               
        a_tags = soup.select("div.sa_text a[href*='https://n.news.naver.com/mnews/']") # 클래스가 sa.text인 div태그, 그 안에 <a>태그를 전부 찾음
        #태그 리스트 순회
        for a_tag in a_tags:
            title = a_tag.text.strip()# <a>태그 안에 있는 텍스트 가져옴, strip은 공백 제거
            link = a_tag['href'] #a_tag에서 기사 제목 부분을 가져옴
            print(title, link) #제목과 링크를 출력함
    else:
        print(f'Error Code = {res.status_code}') #정상 응답이면 200, 아니면 에러코드줌
```

### 2. IMAGE 다운로드

```python
import requests
import os
# 헤더를 넣어야, 핫링크 방지가 됨, user-agent = 내가 브라우저처럼 보이게 하는것
req_header = {
    'referer':'https://comic.naver.com/webtoon/detail?titleId=812354&no=208&week=sun',
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }
# 다운로드 할 이미지 3개의 url 리스트
img_urls = [
    'https://image-comic.pstatic.net/webtoon/812354/208/20251223155029_0d14a9e6cdd6b23342f96d1bd33d01c3_IMAG01_1.jpg',
    'https://image-comic.pstatic.net/webtoon/812354/208/20251223155029_0d14a9e6cdd6b23342f96d1bd33d01c3_IMAG01_2.jpg',
    'https://image-comic.pstatic.net/webtoon/812354/208/20251223155029_0d14a9e6cdd6b23342f96d1bd33d01c3_IMAG01_3.jpg'
]
#리스트를 하나씩 순회한다.
for img_url in img_urls:
    res = requests.get(img_url,headers=req_header) #img_url로 요청 보내고 req_header를 첨부, 응답 상태가 res에 저장
    img_data = res.content #응답 본문을 바이너리 형태로, 이미지는 텍스트 아니라서 res.content 쓰기
    file_name = os.path.basename(img_url) # 파일명만 추출함
    print(file_name)
    with open(file_name, 'wb') as file: #file_name이라는 이름으로 파일 생성 후 열기 wb:바이너리 데이터 쓰기모드
        #지금 어떤 파일을, 몇 바이트로(천단위 콤마) 저장하고 있는지 알려주는 코드
        print(f'Writing to {file_name}({len(img_data):,} bytes)')
        file.write(img_data)
```

## 현재 요청된 페이지의 IMAGE 모두 다운로드하기

```python
import requests
from bs4 import BeautifulSoup
import os

webtoon_url = 'https://comic.naver.com/webtoon/detail?titleId=812354&no=208&week=sun'

req_header = {
    'referer': webtoon_url,
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}
# 웹툰 상세페이지 자체에 get 요청 후 html 받아오기
res = requests.get(webtoon_url, headers=req_header)
if res.ok:
    soup = BeautifulSoup(res.text, 'html.parser') # 받아온 텍스트를 beautifulsoup으로 파싱
    print(len(soup.select("img[src$='.jpg']"))) # src가 .jpg로 끝나는 모든 img 태그 찾기(&: ~로 끝나는)
    print(len(soup.select("img[src*='IMAG01']"))) # src에 IMAG01이 포함된 img 태그 찾기, 위에꺼로 범위 안좁혀져서 다시 범위 좁힘
    img_tags = soup.select("img[src*='IMAG01']") # img_tags 변수에 결과 저장하기

    img_url_list = [] # 이미지 url 담을 리스트 준비하기
    # 기존 방식, img_tags를 하나씩 돌면서 scr의 속성값을 꺼낸후 img_url_list에 추가, [빈 리스트+append]
    for img_tag in img_tags:
        img_url = img_tag['src']
        img_url_list.append(img_url)  
    print(img_url_list[:2]) # 슬라이싱 해서 출력

    #list comprehension 방식
    img_url_list2 = [img_tag['src']for img_tag in img_tags]
    print(img_url_list2[:2])

    imgdir_name = 'img' # 이미지를 저장할 디렉토리 이름을 변수로 정의하기
    if not os.path.isdir(imgdir_name): # img라는 폴더가 존재하는지 확인하고, 없으면 만들기
        os.mkdir(imgdir_name) 

    for img_url in img_url_list2:
        res = requests.get(img_url, headers=req_header)
        img_data = res.content
        # img/파일명.jpg로 추출
        file_path = os.path.join(imgdir_name,os.path.basename(img_url))

        with open(file_path,'wb') as file: # 위 경로로 파일을 바이너리 쓰기 모드로 열기
            print(f'Writing to {file_path}({len(img_data):,} bytes)') # 저장 경로와 파일 크기 출력
            file.write(img_data)  # 이미지 데이터를 실제 파일로 저장하기
```

## 4. 캡처(이미지) API 호출하기

```python
# 캡차 키 발급 요청
import os
import sys
import urllib.request #환경 변수, 시스템 관련 작업으로 import, http 요청 라이브러리

code = "0" # 코드 값으로 어떤 동작을 하는지, 0은 새로운 키 발급하라는 것.
url = f"https://openapi.naver.com/v1/captcha/nkey?code={code}"
request = urllib.request.Request(url) # 요청 객체를 먼저 만든다
request.add_header("X-Naver-Client-Id",CLIENT_ID) #요청 객체에 키 값 추가
request.add_header("X-Naver-Client-Secret",CLIENT_SECRET) #요청 객체에 키 값 추가
response = urllib.request.urlopen(request) #서버에 요청을 전송함
rescode = response.getcode()
if(rescode==200): #상태 코드가 성공(200) 일때만 실행하는 코드
    response_body = response.read() #응답 본문을 바이트 형태로 읽는다
    print(response_body.decode('utf-8')) #바이트를 문자열로 변환한다. 이때 출력시 key 값이 JSON 형태
else:
    print("Error Code:" + rescode) #실패면 에러코드 출력
```
