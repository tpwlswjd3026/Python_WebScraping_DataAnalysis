## 2-1. Nate 뉴스기사 제목 스크래핑하기.

```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from IPython.display import Image, display


section_dict = {'최신뉴스':'100','정치':'200','경제':'300','사회':'400',' 세계':'500','IT/과학':'600'}

req_header = {
    'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
}

def print_news(section_name):
    sid = section_dict.get(section_name,'200')
    if sid == '100':  # 최신뉴스는 recent
        url = f'https://news.nate.com/recent?mid=n0{sid}'
    else:              # 나머지는 section
        url = f'https://news.nate.com/section?mid=n0{sid}'
    print(f'{section_name} 뉴스 {url}')



    res = requests.get(url,headers=req_header)

    if res.ok: 
        html = res.text
        soup = BeautifulSoup(html,'html.parser')

        articles = soup.select("div.mlt01 a.lt1")

        if articles:
            first = articles[0]
            
            title_tag = first.select_one("h2.tit")
            title = title_tag.text.strip() if title_tag else '제목 없음'

            link = urljoin(url, first['href'])

            img_tag = first.select_one("img")
            if img_tag and img_tag.get('src'):
                img_src = urljoin(url, img_tag['src'])
                display(Image(url=img_src))
            else:
                 print('(이미지 없음)')
            print(f'{title} - {link}')
        else:
              print(f'[{section_name}] 기사를 찾을 수 없습니다.')

    else:
        print(f'[{section_name}] Error Code = {res.status_code}')


for section_name in section_dict:
    print_news(section_name)
            
```

## 2-2. 하나의 네이버 웹툰과 1개의 회차에 대한 image 다운로드 하기

```python
import requests
from bs4 import BeautifulSoup
import os

def download_episode(title, no, url):
    req_header = {
        'referer':'https://comic.naver.com/webtoon/detail?titleId=783053&no=250&week=tue',
        'user-agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/144.0.0.0 Safari/537.3'
    }
    imgdir_name = os.path.join('img', title, str(no))

    os.makedirs(imgdir_name, exist_ok=True)

    res = requests.get(url)
    if not res.ok:
        print(f'Error Code = {res.status_code}')
        return

    soup = BeautifulSoup(res.text, 'html.parser')
    img_url_list = [img_tag['src'] for img_tag in soup.select("img[src*='IMAG01']")]

    for img_url in img_url_list:
     res = requests.get(img_url, headers=req_header)
     if res.ok:
       img_data = res.content
       file_path = os.path.join(imgdir_name, os.path.basename(img_url))
       with open(file_path, 'wb') as file:
            print(f'Writing to {file_path} ({len(img_data):,} bytes)')
            file.write(img_data) 
     else:
             print(f'Error Code = {res.status_code} for {img_url}')   

download_episode('김부장', 250,'https://comic.naver.com/webtoon/detail?titleId=783053&no=250&week=tue' )
```
